```python
import numpy as np
filename = 'BlockAssembly_normal_20.csv'

def truncate(x, left, right):
    if (x>right):
        x=right
    elif (x<left):
        x=left
    return round(x,2)        

mean = [100, 120, 80, 100]
std = [20, 30, 10, 15]

with open(filename, 'w') as f:
    f.write('Index,Process1,Process2,Process3,Process4\n')
for i in range(20):
    with open(filename, 'a') as f:
            f.write('%d' % i )
    
    a = np.random.randint(low=1, high=40)
    for j in range(4):
        t = np.random.normal(mean[j],std[j])
        t = truncate(t,mean[j]-1.5*std[j],mean[j]+1.5*std[j]) + a
        with open(filename, 'a') as f:
            f.write(',%.2f' % t)
            
    with open(filename, 'a') as f:
        f.write('\n')
```

```python
import json
info = dict()
data = pd.read_csv('Taillard_1.csv', header=None).to_numpy()
def create_data_with_IAT(idx, lmbda, pt_mean, pt_variability):
    np.random.seed(42)
    for i in range(1): # 1세트 만듦
        info[str(i)] = dict()
        time = 0.0
        iat = 0.0
        for j in range(100):
            if j != 0:
                iat = round(np.random.exponential(lmbda))
                time += iat
            pt_1 = data[j].tolist()
            pt_high = pt_mean + pt_variability
            pt_low = pt_mean - pt_variability
            pt_2 = [np.random.randint(low=pt_low, high=pt_high, size=1).repeat(4).tolist()]
            d = {
                'Name':'Job_'+str(j),
                'Created':time,
                'IAT':iat,
                'Work':['FS','PMS'],
                'num_process':[5,1],
                'type_machine':['Unrelated', 'Identical'],
                'num_machine': [[1, 1, 1, 1, 1], [4]],
                'processing_time':[pt_1,pt_2],
            }
            info[str(i)]['Job_'+str(j)]=d
    # _data.json 파일에 딕셔너리 형태로 info 저장
    with open("data_Taillard_{0}.json".format(str(idx)), 'w') as f:
        json.dump(info, f, indent='\t')
```

```python
create_data_with_IAT(6, 80, 160, 10)
create_data_with_IAT(7, 80, 160, 10)
create_data_with_IAT(8, 80, 160, 10)
```

```python
def calculate_makespan(blocks, sequence):
    if isinstance(sequence, list):
        sequence = np.array(sequence)
        
    num_of_blocks = blocks.shape[0]
    num_of_process = blocks.shape[1]
    temp = np.zeros((num_of_blocks + 1, num_of_process + 1))
    for i in range(1, num_of_blocks + 1):
        for j in range(1, num_of_process + 1):
            if temp[i - 1, j] > temp[i, j - 1]:
                temp[i, j] = temp[i - 1, j] + blocks[sequence[i - 1], j - 1]
            else:
                temp[i, j] = temp[i, j - 1] + blocks[sequence[i - 1], j - 1]
    C_max = temp[num_of_blocks, num_of_process]
    return C_max
```

```python
blocks = np.array([[3,5,8],
                   [4,7,5],
                   [2,6,9]])
sequence = np.array([0,1,2])
calculate_makespan(blocks, sequence)
```
